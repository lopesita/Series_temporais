1. Séries Temporais

Uma série temporal é definida como uma sequência ordenada de observações de uma variável contínua ou discreta, coletadas em intervalos de tempo regulares e sucessivos (Brockwell; Davis, 2016). No contexto analítico deste projeto, focado no mercado de combustíveis líquidos com base nos dados da ANP, a variável de interesse $Y_t$ representa o volume de vendas de óleo diesel com destino ao estado de São Paulo no instante $t$.

A literatura concorda que a análise exploratória fundamental dessas séries exige a compreensão de sua estrutura interna por meio da decomposição (Montgomery; Jennings; Kulahci, 2015).A decomposição clássica postula que a série é gerada pela combinação de três componentes inobserváveis (Morettin; Toloi, 2006; Peixeiro, 2022):

  * Tendência ($T_t$): A direção e o comportamento de longo prazo da série (crescimento, declínio ou estabilidade).

  * Sazonalidade ($S_t$): Padrões intra-anuais que se repetem em intervalos fixos e conhecidos, frequentemente associados a ciclos econômicos, climáticos ou safras agrícolas.

  * Ruído ou Componente Aleatório ($R_t$):
  A variação residual, irregular e estocástica não explicada pelos outros componentes.A relação empírica entre esses componentes pode ser modelada de forma aditiva, recomendada quando a amplitude da variação sazonal é independente do nível da série:$$Y_t = T_t + S_t + R_t$$Ou de forma multiplicativa, preferível quando a variação sazonal cresce proporcionalmente à tendência (Montgomery; Jennings; Kulahci, 2015):

  $$Y_t = T_t \times S_t \times R_t$$



2.    Estacionariedade   
Na literatura estatística e econométrica, trabalha-se predominantemente com o conceito de estacionariedade fraca (ou estacionariedade de segunda ordem). Para que uma série temporal $Y_t$ seja classificada como fracamente estacionária, ela deve satisfazer simultaneamente três condições matemáticas (Enders, 2014):

  * Média constante no tempo:
  O valor esperado (média) da série é invariável para qualquer instante $t$.
  $$E(Y_t) = \mu$$
  
  * Variância constante no tempo (homocedasticidade):
  A dispersão ou volatilidade dos dados em torno da média não aumenta nem diminui com o passar do tempo. $$Var(Y_t) = \sigma^2$$
  
  * Autocovariância dependente apenas da defasagem ($k$): A relação linear entre duas observações depende estritamente do intervalo de tempo ($k$) que as separa (seus lags), e não do momento histórico $t$ exato em que foram registradas.$$Cov(Y_t, Y_{t-k}) = \gamma_k$$


3.    Autocorrelação (ACF) e Autocorrelação Parcial (PACF)

A autocorrelação mede o grau de dependência linear entre as observações de uma mesma série temporal em diferentes defasagens (lags). Ela é fundamental para entender a inércia do consumo ao longo do tempo. A Função de Autocorrelação (ACF) para um lag $k$, denotada por $r_k$, é calculada pela fórmula:$$r_k = \frac{\sum_{t=1}^{N-k} (Y_t - \bar{Y})(Y_{t+k} - \bar{Y})}{\sum_{t=1}^N (Y_t - \bar{Y})^2}$$Onde $\bar{Y}$ é a média da série e $N$ é o número total de observações.Complementarmente, a Função de Autocorrelação Parcial (PACF) mede a correlação entre $Y_t$ e $Y_{t-k}$ removendo o efeito linear das defasagens intermediárias (de $t-1$ até $t-k+1$). Ambas as funções (ACF e PACF) são as principais ferramentas gráficas e estatísticas utilizadas para identificar a ordem dos processos autorregressivos e de médias móveis em modelos estatísticos.


4.   AutoARIMA

 O algoritmo AutoARIMA, desenvolvido por Hyndman e Khandakar (2008), automatiza a complexa tarefa de definir os parâmetros $p$, $d$ e $q$. Ele realiza testes de raiz unitária para definir a diferenciação ($d$) e, em seguida, realiza uma busca no espaço de hiperparâmetros para $p$ e $q$, selecionando o modelo que minimiza o Critério de Informação de Akaike (AIC), penalizando a complexidade excessiva:$$AIC = -2 \log(L) + 2k$$Onde $L$ é a verossimilhança do modelo e $k$ é o número de parâmetros estimados.


5. XGBoost

A previsão de séries temporais passou por uma transição metodológica com a ascensão das técnicas de Aprendizado de Máquina, que superam a premissa de linearidade dos modelos estocásticos tradicionais (Bontempi; Taieb; Le Borgne, 2012; Makridakis; Spiliotis; Assimakopoulos, 2018).

O XGBoost (eXtreme Gradient Boosting), proposto por Chen e Guestrin (2016), é um algoritmo de aprendizado supervisionado baseado em um conjunto (ensemble) de árvores de decisão. Para aplicá-lo a séries temporais, os dados sequenciais são reestruturados matricialmente, onde as defasagens (lags) assumem o papel de variáveis explicativas (features). O algoritmo constrói $K$ árvores de regressão sequenciais, onde cada nova árvore é treinada para corrigir os erros residuais das árvores anteriores (otimização por gradiente). A previsão final $\hat{y}_i$ é dada por:$$\hat{y}_i = \sum_{k=1}^K f_k(x_i), \quad f_k \in \mathcal{f}$$ treinamento ocorre mediante a minimização de uma Função Objetivo rigorosa, composta pela perda de previsão $l$ e um termo de regularização $\Omega$ que controla a complexidade estrutural das árvores e previne o overfitting (Nielsen, 2016):$$\text{Obj} = \sum_{i=1}^n l(y_i, \hat{y}_i) + \sum_{k=1}^K \Omega(f_k)$$A função de penalidade de regularização $\Omega$ para uma árvore individual $f$ é formulada como:$$\Omega(f) = \gamma T + \frac{1}{2} \lambda \sum_{j=1}^T w_j^2$$Onde $T$ representa o número de folhas da árvore, $w_j$ o escore (peso) numérico de cada folha, enquanto $\gamma$ e $\lambda$ são hiperparâmetros de penalização (Chen; Guestrin, 2016). A aplicação do XGBoost é altamente justificada em cenários de demanda energética, pois ele consegue extrair mapeamentos não lineares complexos e assimilar choques assimétricos no mercado de combustíveis com maior precisão do que modelos puramente lineares como o ARIMA (Makridakis; Spiliotis; Assimakopoulos, 2018).





# **Pipeline da Solução**

Para a realização deste estudo, foi estruturado um pipeline de dados focado em experimentação e análise comparativa, executado integralmente em ambiente cloud através da plataforma Google Colab. O fluxo de trabalho divide-se em etapas sequenciais que garantem a integridade dos dados, desde a ingestão bruta até a validação dos modelos preditivos.


1.   Ingestão e Limpeza dos Dados
      
      A etapa inicial compreende a extração dos dados brutos a partir do conjunto de dados Liquidos_Vendas_Atual, fornecido pela ANP em formato CSV.

      *   Filtragem Estratégica: seleção do produto diesel e do estado de São Paulo como destino, delimitando o escopo ao problema de negócio proposto.
      *   Tratamento Temporal: conversão e formatação das colunas de mês e ano para o tipo datetime, permitindo a ordenação cronológica necessária para séries temporais.
      *   Saneamento: identificação e tratamento de possíveis valores ausentes ou inconsistências nas unidades de medida (mil m³).


2.   Análise Exploratória de Dados

      Esta etapa consiste na investigação estatística da série temporal para identificar padrões subjacentes e validar as premissas exigidas pelos algoritmos. O foco reside em três pilares:

      *   Decomposição da série em seus componentes fundamentais: Tendência (direção de longo prazo), Sazonalidade (padrões cíclicos de consumo de diesel) e Resíduos (ruído branco ou variações aleatórias). Essa separação permite entender se a performance do modelo será guiada por padrões repetitivos ou por mudanças estruturais de mercado.
      *   Diagnóstico de Estacionariedade: Aplicação de testes estatísticos rigorosos, como o Teste de Dickey-Fuller Aumentado (ADF). O objetivo é verificar se as propriedades estatísticas da série (média e variância) são constantes no tempo, critério mandatório para a viabilidade de modelos como o autoARIMA
      *   Análise de Autocorrelação (ACF e PACF): Utilização das funções de autocorrelação e autocorrelação parcial para identificar a dependência serial dos dados. Esta análise é o "termômetro" para a etapa de Engenharia de Atributos, determinando quantos períodos de defasagem (lags) são estatisticamente relevantes para prever o volume futuro.

3.   Engenharia de Atributos
      
      Antes da modelagem, os dados tratados passam por transformações para capturar padrões de sazonalidade e tendência. Serão criadas variáveis de defasagem (lags) e/ou médias móveis, que auxiliam os modelos a compreender a dependência serial dos dados históricos de vendas de diesel, além de preparar a estrutura necessária para os algoritmos de aprendizado de máquina.

4.   Desenvolvimento e Comparação de Algoritmos

      O núcleo do pipeline reside na modelagem preditiva, onde o ambiente Colab é utilizado para a construção de dois paradigmas distintos:

      *   Modelagem Estatística Clássica: implementação de modelos como autoARIMA, focados na estrutura estocástica da série.
      *   Modelagem via Aprendizado de Máquina: implementação de algoritmos supervisionados (como XGBoost ou Random Forest), capazes de capturar relações não lineares complexas.

5.   Avaliação e Validação de Performance

      Dado que o projeto não prevê orquestramento ou produtização imediata, a etapa final do pipeline é a Avaliação de Performance.
      A métrica de sucesso será definida pelo erro médio, utilizando indicadores como MAE (Mean Absolute Error) e/ou RMSE (Root Mean Square Error).
