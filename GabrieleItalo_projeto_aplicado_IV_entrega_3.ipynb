### **Análise Bruta**

A avaliação inicial da série de vendas foi realizada por meio do teste de Dickey-Fuller Aumentado (ADF), além da inspeção visual das funções de autocorrelação (ACF) e autocorrelação parcial (PACF). O objetivo dessa etapa é verificar se a série apresenta estacionariedade, isto é, se suas propriedades estatísticas, como média e variância, permanecem relativamente constantes ao longo do tempo.

O resultado do teste ADF apresentou estatística de 0,2654 e p-valor de 0,9757. Como o p-valor é muito superior ao nível de significância usual de 5% (0,05), não se rejeita a hipótese nula de presença de raiz unitária. Em termos práticos, isso indica que a série não é estacionária. Esse resultado sugere que a série bruta contém componentes sistemáticos, como tendência, sazonalidade ou ambos, que impedem sua estabilidade ao longo do tempo.

A análise da função de autocorrelação reforça esse diagnóstico. Observa-se que os primeiros lags apresentam autocorrelações elevadas, especialmente no lag 1, o que indica forte dependência temporal entre valores consecutivos. Além disso, a ACF não decai rapidamente para zero, comportamento típico de séries não estacionárias. Também se destacam picos expressivos em defasagens mais longas, sobretudo em torno do lag 12 e seus arredores, o que sugere a existência de sazonalidade. Em séries mensais, por exemplo, esse padrão costuma indicar repetição anual do comportamento observado.

Já a função de autocorrelação parcial mostra um pico bastante acentuado no lag 1, sinalizando que o passado imediato exerce influência direta e relevante sobre os valores atuais da série. Há ainda sinais em lags sazonais, principalmente próximos de 12, o que reforça a hipótese de uma estrutura sazonal além da dependência de curto prazo. Esse padrão é compatível com séries que não podem ser modeladas adequadamente em sua forma bruta, exigindo transformações prévias.

Do ponto de vista metodológico, os resultados obtidos indicam que a série precisa passar por procedimentos de transformação antes da modelagem preditiva. Entre as abordagens mais adequadas estão a diferenciação simples, para remover tendência, e a diferenciação sazonal, para tratar a repetição periódica observada. Após essas transformações, recomenda-se reaplicar o teste ADF e reavaliar os gráficos de ACF e PACF, a fim de verificar se a série passou a atender ao pressuposto de estacionariedade.

Em síntese, a análise estatística da série bruta evidencia que ela apresenta forte persistência temporal, não estacionariedade e indícios claros de sazonalidade. Portanto, a utilização da série em estado bruto em modelos autorregressivos pode comprometer a qualidade das estimativas e previsões. Assim, a etapa de transformação da série se mostra indispensável para a construção de um modelo temporal mais robusto e confiável.

```python
# Teste Estacionariedade
print("--- Teste de Estacionaridade (ADF) ---")
resultado_adf = adfuller(serie_vendas)
print(f'Estatística ADF: {resultado_adf[0]:.4f}')
print(f'p-valor: {resultado_adf[1]:.4f}')

if resultado_adf[1] <= 0.05:
    print("Resultado: A série é estacionária.")
else:
    print("Resultado: A série não é estacionária.")
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))

# ACF - sazonalidade (lags distantes)
plot_acf(serie_vendas, lags=24, ax=ax1, title="Autocorrelação (ACF) - Série Bruta")

# PACF - passado imediato (lags próximos)
plot_pacf(serie_vendas, lags=24, ax=ax2, title="Autocorrelação Parcial (PACF) - Série Bruta")

plt.tight_layout()
plt.show()
```

### **Análise Decomposta**

A decomposição da série temporal mostra que os dados são formados por três componentes bem definidos: tendência, sazonalidade e resíduo. A componente de tendência apresenta crescimento gradual ao longo do período, indicando aumento no nível médio das vendas com o passar do tempo.

A componente sazonal revela um padrão cíclico bastante regular, com repetições em intervalos constantes, o que confirma a existência de sazonalidade na série. Esse comportamento já era sugerido anteriormente pela ACF, especialmente nas defasagens mais longas.

Já os resíduos oscilam em torno de zero, sem um padrão tão evidente quanto os demais componentes, o que indica que boa parte da estrutura da série foi explicada pela tendência e pela sazonalidade. Além disso, após a diferenciação, o teste ADF apresentou p-valor de 0,0035, mostrando que a série se tornou estacionária.

Assim, a análise decomposta confirma que a série original possuía tendência crescente e forte sazonalidade, e que a diferenciação foi adequada para remover a tendência e tornar a série apropriada para modelagem estatística.

```python
dec = seasonal_decompose(serie_vendas, model='additive', period=12)

fig, (ax1, ax2, ax3, ax4) = plt.subplots(4, 1, figsize=(15, 12), sharex=True)
dec.observed.plot(ax=ax1, title='Original', color = 'red')
dec.trend.plot(ax=ax2, title='Tendência', color = 'red')
dec.seasonal.plot(ax=ax3, title='Sazonalidade', color = 'red')
dec.resid.plot(ax=ax4, title='Resíduo', color = 'red')
plt.tight_layout()
plt.show()
```

```python
serie_diff = serie_vendas.diff().dropna()

print("-Teste ADF após Diferenciação-")
res_diff = adfuller(serie_diff)
print(f'Novo p-valor: {res_diff[1]:.4f}')

if res_diff[1] <= 0.05:
    print("Resultado: A série é estacionária após diferenciação.")
```

Os gráficos de ACF e PACF da série diferenciada indicam que a diferenciação foi eficaz para reduzir a tendência e tornar a série mais estável. Na ACF, a maior parte das autocorrelações permanece dentro do intervalo de confiança, o que sugere diminuição da dependência serial observada na série bruta. Ainda assim, alguns picos em defasagens específicas, como em torno do lag 12 e lag 24, apontam que a sazonalidade não foi totalmente eliminada.

Na PACF, também se observa que a maioria dos coeficientes está próxima da faixa de confiança, embora alguns lags permaneçam relevantes. Esse comportamento mostra que, após a diferenciação simples, a série perdeu grande parte da tendência, mas ainda mantém traços de dependência sazonal.

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))
plot_acf(serie_diff, lags=24, ax=ax1, title="ACF - Série Diferenciada (Sem Tendência)")
plot_pacf(serie_diff, lags=24, ax=ax2, title="PACF - Série Diferenciada (Sem Tendência)")

plt.tight_layout()
plt.show()
```

## **XGBoost**

O pré-processamento dos dados foi estruturado para converter a série temporal bruta em um formato de aprendizado supervisionado, permitindo que o algoritmo XGBoost capture a dependência serial dos dados. Esta etapa é crítica, pois a qualidade dos inputs determina a capacidade de generalização do modelo frente às volatilidades da demanda de diesel em São Paulo.

```python
display(df_new)
```

Inicialmente, realizou-se o agrupamento multidimensional das variáveis temporais e geográficas. A base foi consolidada pela soma da quantidade de produto (mil m³), utilizando como chaves de agrupamento o ano, mês, nome do produto e as unidades federativas de destino.

```python
colunas_agrupamento = ['Ano', 'Mês', 'Nome do Produto', 'Região Origem', 'UF Destino']
df_model = df_new.groupby(colunas_agrupamento, as_index=False).agg({'Quantidade de Produto (mil m³)': 'sum'})
print(df_model)
```

Para isolar o fenômeno de interesse, os dados foram pivotados em uma estrutura matricial onde as regiões de origem tornaram-se colunas explicativas, e a vacância de dados foi tratada via input de valor zero (fillna(0)), assumindo a ausência de transações comerciais nesses períodos específicos.

```python
# Criar a série temporal pivotada
df_ts = df_model.pivot_table(index=['Ano', 'Mês'],
                             columns='Região Origem',
                             values='Quantidade de Produto (mil m³)',
                             aggfunc='sum').fillna(0)

# Criar o alvo: Soma total que chega em SP
df_ts['Total_SP'] = df_ts.sum(axis=1)
```

Dada a natureza estocástica das séries temporais de combustíveis, aplicou-se o teste estatístico de Dickey-Fuller Aumentado (ADF) para verificar a estacionariedade da variável alvo (Total_SP).
*   Diagnóstico Inicial: A série original apresentou indícios de não-estacionariedade (presença de tendência), o que pode comprometer a performance de algoritmos baseados em árvores.

*   Diferenciação: Para mitigar esse efeito, aplicou-se a técnica de diferenciação de primeira ordem ($y_t' = y_t - y_{t-1}$), estabilizando a média da série. A eficácia da transformação foi validada por um novo teste ADF, que confirmou a estacionariedade da série diferenciada (p-valor $\leq$ 0,05).



```python
# Teste estacionariedade nova série
print("--- Teste de Estacionaridade (ADF) para 'Total_SP'---")
resultado_adf = adfuller(df_ts['Total_SP'])
print(f'Estatística ADF: {resultado_adf[0]:.4f}')
print(f'p-valor: {resultado_adf[1]:.4f}')

if resultado_adf[1] <= 0.05:
    print("Resultado: A série 'Total_SP' é estacionária.")
else:
    print("Resultado: A série 'Total_SP' não é estacionária.")
```

```python
# Aplicar a diferenciação
df_diff = df_ts.diff().dropna()
```

```python
# Teste após aplicação da diferenciação
print("--- Teste de Estacionaridade (ADF) para 'Total_SP'---")
resultado_adf = adfuller(df_diff['Total_SP'])
print(f'Estatística ADF: {resultado_adf[0]:.4f}')
print(f'p-valor: {resultado_adf[1]:.4f}')

if resultado_adf[1] <= 0.05:
    print("Resultado: A série 'Total_SP' é estacionária.")
else:
    print("Resultado: A série 'Total_SP' não é estacionária.")
```

```python
#Transformar Ano/Mês em um único índice temporal para facilitar os lags
df_model['Data'] = pd.to_datetime(df_model['Ano'].astype(str) + '-' + df_model['Mês'].astype(str) + '-01')
```

```python
# Pivotar para ter as regiões como colunas
df_pivot = df_model.pivot_table(index='Data',
                                columns='Região Origem',
                                values='Quantidade de Produto (mil m³)',
                                aggfunc='sum').fillna(0)

df_pivot['Total_SP'] = df_pivot.sum(axis=1)

# Criar a Matriz de Features (X)
df_features = pd.DataFrame(index=df_pivot.index)
df_features['target'] = df_pivot['Total_SP']
```

Para que o XGBoost opere sobre dados temporais, foi necessário criar uma "memória" no conjunto de dados. Para isso, implementou-se a engenharia de atributos autorregressivos (Lags). Foram criadas janelas de atraso (Total_SP_lag1 e Total_SP_lag2), deslocando os valores da série em um e dois meses, respectivamente.

```python
# Criar Lags Autorregressivos
df_features['Total_SP_lag1'] = df_pivot['Total_SP'].shift(1)
df_features['Total_SP_lag2'] = df_pivot['Total_SP'].shift(2)
```

Essa abordagem transforma o valor histórico em uma feature preditora, permitindo que o modelo identifique padrões de autocorrelação. Após a criação dos lags, realizou-se o tratamento de descontinuidade através da remoção dos registros iniciais, resultando em uma matriz de atributos ($X$) e um vetor de alvo ($y$) prontos para a fase de treinamento.

```python
# Remover os primeiros 6 meses que ficaram com NaN devido a criação do lag
df_final = df_features.dropna()
```

```python
# Definir Variáveis de Treino e Alvo
X = df_final.drop(columns=['target'])
y = df_final['target']

print("Colunas criadas para o XGBoost:")
print(X.columns.tolist())
```

# **Modelo base**

## **XGBoost**

Nesta etapa, o estudo avançou para a fase de modelagem preditiva utilizando o algoritmo XGBoost, um método de Gradient Boosting amplamente reconhecido pela sua eficiência em tarefas de regressão com dados tabulares. O foco desta versão inicial foi estabelecer uma linha de base (baseline) utilizando exclusivamente atributos autorregressivos.

A base de dados foi segmentada utilizando o critério de divisão cronológica, com o marco temporal definido em 1º de janeiro de 2025. Dados anteriores a este período compuseram o conjunto de treinamento

```python
# Split treino e teste
split_date = '2025-01-01'

X_train = X[X.index < split_date]
y_train = y[y.index < split_date]

X_test = X[X.index >= split_date]
y_test = y[y.index >= split_date]

print(f"Treino: {len(X_train)} meses | Teste: {len(X_test)} meses")
```

O modelo foi configurado com 500 estimadores e uma taxa de aprendizado (learning rate) de 0,05, visando uma convergência estável. Para prevenir o overfitting (sobreajuste), aplicou-se uma profundidade máxima de 5 níveis para as árvores e uma taxa de subamostragem de 80% das observações e colunas por iteração. A função de perda utilizada foi o erro quadrático médio

```python
# criando o modelo
model = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    objective='reg:squarederror',
    random_state=42
)
```

```python
# Treino
model.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_test, y_test)],
    verbose=False
)
```

```python
# teste
y_pred = model.predict(X_test)
```

O coeficiente de determinação ($R^2$) negativo é um indicador técnico relevante: ele demonstra que as predições do modelo foram, em média, menos precisas do que a simples utilização da média histórica da série. O MAE de 122,27 mil m³ indica um desvio significativo em relação ao volume real movimentado, sugerindo que o modelo não conseguiu capturar as variações sazonais e os choques de demanda do estado de São Paulo no período de 2025.

```python
# Métricas
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE: {mae:.2f} mil m³")
print(f"RMSE: {rmse:.2f} mil m³")
print(f"R² Score: {r2:.4f}")
```

A análise do gráfico de comparação entre os valores reais e preditos confirma visualmente a baixa aderência do modelo. Observou-se um fenômeno de subestimação sistemática no segundo semestre de 2025 e um erro de fase (atraso nas respostas) nas flutuações de curto prazo.

Portanto, os resultados desta V1 validam a necessidade técnica da V2, que buscará a otimização do pré-processamento através da inclusão de lags de maior ordem, médias móveis e outras varíaveis preditoras.

## **Auto Arima**

O pré-processamento dos dados para o Auto ARIMA foi estruturado com o objetivo de isolar e organizar a série temporal correspondente às vendas de Diesel B destinadas ao estado de São Paulo, que constitui o objeto central de análise deste estudo. Para isso, a base original foi filtrada pelas variáveis Nome do Produto e UF Destino, mantendo apenas os registros referentes ao Diesel B com destino a São Paulo. Em seguida, os volumes foram agregados mensalmente a partir das colunas Ano e Mês, utilizando a variável Quantidade de Produto (mil m³) como medida de interesse. A partir dessa série univariada, foram conduzidas as etapas diagnósticas e de modelagem estatística, incluindo análise gráfica, teste de estacionariedade e geração de previsões.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

serie_vendas = (
    df_model.groupby(['Ano', 'Mês'])['Quantidade de Produto (mil m³)']
    .sum()
    .reset_index()
)

serie_vendas['Data'] = pd.to_datetime(
    serie_vendas['Ano'].astype(str) + '-' + serie_vendas['Mês'].astype(str) + '-01'
)

serie_vendas.set_index('Data', inplace=True)
serie_vendas = serie_vendas['Quantidade de Produto (mil m³)'].sort_index().asfreq('MS')

print(serie_vendas.head())
print(serie_vendas.index.freq)
```

O teste de Dickey-Fuller Aumentado aplicado à série original apresentou estatística ADF de 0,2654 e p-valor de 0,9757, indicando que a hipótese nula de presença de raiz unitária não pode ser rejeitada. Isso mostra que a série original não é estacionária, ou seja, apresenta comportamento temporal incompatível com os pressupostos básicos exigidos para a modelagem ARIMA em sua forma bruta.


```python
#Teste de estacionariedade

from statsmodels.tsa.stattools import adfuller

resultado_adf = adfuller(serie_vendas.dropna())

print(f'Estatística ADF: {resultado_adf[0]:.4f}')
print(f'p-valor: {resultado_adf[1]:.4f}')
```

Após a aplicação da diferenciação simples, o teste foi realizado novamente e apresentou estatística ADF de -3,7429 e p-valor de 0,0035. Como esse valor é inferior ao nível de significância de 5%, rejeita-se a hipótese nula, indicando que a série diferenciada se tornou estacionária.

Dessa forma, os resultados mostram que a diferenciação foi eficaz para remover a não estacionariedade da série, tornando os dados mais adequados para a aplicação do modelo Auto ARIMA na projeção das vendas de Diesel B em São Paulo.

```python
serie_diff = serie_vendas.diff().dropna()

resultado_adf_diff = adfuller(serie_diff)

print(f'Estatística ADF da série diferenciada: {resultado_adf_diff[0]:.4f}')
print(f'p-valor da série diferenciada: {resultado_adf_diff[1]:.4f}')
```

Para a modelagem preditiva, a série temporal foi dividida em dois subconjuntos: treino e teste. O conjunto de treino ficou composto por 96 meses, enquanto o conjunto de teste reuniu os 12 meses finais da série. Essa separação foi realizada preservando a ordem cronológica dos dados, o que é essencial em problemas de séries temporais, já que o modelo deve aprender com observações passadas para projetar valores futuros.

A escolha de reservar os últimos 12 meses para teste permite avaliar o desempenho do Auto ARIMA em um horizonte equivalente a um ano completo, o que é adequado para séries mensais e útil para verificar a capacidade do modelo de reproduzir o comportamento recente das vendas de Diesel B em São Paulo. Dessa forma, o conjunto de teste funciona como uma validação fora da amostra, permitindo analisar a qualidade preditiva do modelo em dados não utilizados no ajuste.

```python
train = serie_vendas.iloc[:-12]
test = serie_vendas.iloc[-12:]

print(f'Treino: {len(train)} meses')
print(f'Teste: {len(test)} meses')
```

O ajuste do Auto ARIMA foi realizado com busca automática dos parâmetros, considerando a sazonalidade mensal da série, com período igual a 12. Durante o processo, o algoritmo testou diferentes combinações de componentes autorregressivos, de diferenciação, de média móvel e de termos sazonais, selecionando como melhor especificação o modelo ARIMA(0,1,1)(2,0,0)[12], com base no menor valor do AIC = 1074,736.

Esse resultado indica que a série precisou de uma diferenciação não sazonal (d = 1) para atingir comportamento mais estável, o que está de acordo com o teste de estacionariedade realizado anteriormente. Além disso, o modelo selecionado incorporou um termo de média móvel não sazonal e dois termos autorregressivos sazonais, relacionados às defasagens de 12 e 24 meses, evidenciando que o comportamento das vendas apresenta influência de padrões anuais recorrentes.

Os coeficientes estimados foram estatisticamente significativos, com p-valores inferiores a 0,05, o que sugere contribuição relevante desses componentes para a explicação da dinâmica da série. Em especial, os termos sazonais nas defasagens 12 e 24 reforçam a presença de dependência temporal associada à sazonalidade das vendas de Diesel B em São Paulo.

De forma complementar, o teste Ljung-Box apresentou p-valor de 0,60, indicando ausência de autocorrelação residual significativa no primeiro lag. Isso sugere que o modelo foi capaz de capturar adequadamente a estrutura serial principal da série. Assim, o Auto ARIMA selecionado mostrou-se estatisticamente consistente para a etapa de projeção das vendas.

```python
!pip install pmdarima -q

from pmdarima import auto_arima

modelo_auto = auto_arima(
    train,
    seasonal=True,
    m=12,
    d=None,
    D=None,
    trace=True,
    error_action='ignore',
    suppress_warnings=True,
    stepwise=True,
    information_criterion='aic'
)

print(modelo_auto.summary())
```

Observa-se que o modelo conseguiu reproduzir, de maneira geral, o comportamento das vendas no período projetado, acompanhando a direção predominante da série real. As previsões se mantêm próximas aos valores observados em boa parte dos meses, o que indica que o modelo foi capaz de captar parte importante da dinâmica temporal presente na série, especialmente seu padrão sazonal e sua dependência em relação aos períodos anteriores.

A faixa de confiança também desempenha papel relevante na interpretação dos resultados, pois expressa a incerteza inerente ao processo de previsão. Nota-se que os valores reais permanecem próximos ou dentro desse intervalo em diversos momentos, o que sugere consistência estatística nas projeções realizadas. Ao mesmo tempo, a amplitude da faixa mostra que, embora o modelo apresente boa capacidade de acompanhamento da trajetória da série, existe variabilidade associada às estimativas, especialmente nos períodos mais à frente do horizonte projetado.

```python
previsoes = pd.Series(
    modelo_auto.predict(n_periods=len(test)),
    index=test.index
)

print(previsoes)
```

```python
previsao, conf_int = modelo_auto.predict(n_periods=len(test), return_conf_int=True)

previsoes = pd.Series(previsao, index=test.index)
conf_int_df = pd.DataFrame(conf_int, index=test.index, columns=['limite_inferior', 'limite_superior'])

plt.figure(figsize=(14, 6))

plt.plot(train.index, train, label='Treino')
plt.plot(test.index, test, label='Real')
plt.plot(previsoes.index, previsoes, label='Previsão Auto ARIMA')

plt.fill_between(
    conf_int_df.index,
    conf_int_df['limite_inferior'],
    conf_int_df['limite_superior'],
    alpha=0.2,
    label='Intervalo de confiança'
)

plt.title('Projeção das vendas de Diesel B em São Paulo com Auto ARIMA')
plt.xlabel('Data')
plt.ylabel('Quantidade de produto (mil m³)')
plt.legend()
plt.grid(True)
plt.show()
```

De modo geral, o gráfico sugere que o Auto ARIMA apresentou desempenho satisfatório na projeção das vendas de Diesel B em São Paulo, mostrando aderência razoável entre os valores previstos e os valores efetivamente observados. Assim, a análise gráfica reforça os resultados estatísticos já obtidos nas etapas anteriores, indicando que o modelo se mostrou adequado para representar o comportamento recente da série e apoiar a projeção de vendas em horizonte de curto prazo.
